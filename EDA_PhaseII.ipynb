T-Test

```python
group0 = data[data['morekids'] == 0]['age']
group1 = data[data['morekids'] == 1]['age']

t_stat, p_value = stats.ttest_ind(
    group0,
    group1,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("There is a statistically significant difference between the two groups.")
else:
    print("There is no statistically significant difference between the two groups.")
```

K-Means Clustering

```python
cluster_data = data[['age', 'work', 'morekids_numeric']].dropna().copy()

print(cluster_data.shape)
```

Standardization

```python
print("Inspecting cluster_data before standardization:")
print(cluster_data.info())
print("\nFirst 5 rows of cluster_data:")
print(cluster_data.head())

# Explicitly ensure all columns are numeric, coercing errors
for col in cluster_data.columns:
    cluster_data[col] = pd.to_numeric(cluster_data[col], errors='coerce')
cluster_data.dropna(inplace=True) # Drop rows where coercion resulted in NaN

print("\nInspecting cluster_data after type coercion and NaN removal:")
print(cluster_data.info())
print("\nFirst 5 rows of cluster_data:")
print(cluster_data.head())

scaler = StandardScaler()
scaled_data = scaler.fit_transform(cluster_data)

print("\nScaled data first 5 rows:")
print(scaled_data[:5])
```

### Determining the Optimal Number of Clusters

#### Elbow Method (WCSS)

```python
from sklearn.cluster import KMeans

wcss = []
max_clusters = 10

for i in range(1, max_clusters + 1):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(scaled_data)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(range(1, max_clusters + 1), wcss, marker='o', linestyle='--')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Within-Cluster Sum of Squares (WCSS)')
plt.xticks(range(1, max_clusters + 1))
plt.grid(True)
plt.show()
```

#### Silhouette Score

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

# Silhouette Score requires at least 2 clusters and up to n_samples - 1
# We'll evaluate from 2 to max_clusters
for i in range(2, max_clusters + 1):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(scaled_data)
    score = silhouette_score(scaled_data, kmeans.labels_)
    silhouette_scores.append(score)

plt.figure(figsize=(10, 6))
plt.plot(range(2, max_clusters + 1), silhouette_scores, marker='o', linestyle='--')
plt.title('Silhouette Score For Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.xticks(range(2, max_clusters + 1))
plt.grid(True)
plt.show()
```

K-Means with K = 3

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(scaled_data)

cluster_data['KMeans_Cluster'] = clusters

print(cluster_data.head())
```

Cluster Sizes

```python
print(cluster_data['KMeans_Cluster'].value_counts())
```

Cluster Centers

```python
print("Cluster Centers:")
print(kmeans.cluster_centers_)
```

```python
original_centers = scaler.inverse_transform(
    kmeans.cluster_centers_
)

centers_df = pd.DataFrame(
    original_centers,
    columns=cluster_data.drop(columns=['KMeans_Cluster']).columns
)

print(centers_df)
```

Visualize K-Means

```python
plt.figure(figsize=(8,5))

plt.scatter(
    cluster_data['age'],
    cluster_data['morekids_numeric'], # Corrected column name
    c=cluster_data['KMeans_Cluster']
)

plt.title("K-Means Clustering: Age vs More Kids (Numeric)")
plt.xlabel("Age")
plt.ylabel("More Kids (Numeric)")

plt.show()
```

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot for data points
scatter = ax.scatter(
    cluster_data['age'],
    cluster_data['work'],
    cluster_data['morekids_numeric'],
    c=cluster_data['KMeans_Cluster'],
    cmap='viridis', # Colormap for clusters
    s=50, # Size of points
    alpha=0.6 # Transparency
)

# Scatter plot for cluster centers
# Make sure to use the correct columns for centers_df
ax.scatter(
    centers_df['age'],
    centers_df['work'],
    centers_df['morekids_numeric'],
    marker='X', # Marker for centers
    s=200,      # Size of center markers
    c='red',    # Color of center markers
    label='Cluster Centers'
)

ax.set_title('3D K-Means Clustering: Age, Work, and More Kids (Numeric)')
ax.set_xlabel('Age')
ax.set_ylabel('Work')
ax.set_zlabel('More Kids (Numeric)')

ax.legend()
plt.show()
```

### Interpretation of Cluster Centers

Now that we have performed K-Means clustering and obtained the cluster centers in their original scale, we can interpret the characteristics of each cluster. The `centers_df` DataFrame provides the mean values for 'age', 'work', and 'morekids_numeric' for each of the three clusters. We will use these values to describe what each cluster represents.

Silhouette Score

```python
sil_score = silhouette_score(
    scaled_data,
    clusters
)

print("Silhouette Score:", sil_score)
```

Hierarchical Clustering

```python
linkage_matrix = linkage(
    scaled_data,
    method='single',
    metric='euclidean'
)
```

Dendrogram

```python
plt.figure(figsize=(12,6))

dendrogram(linkage_matrix)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")

plt.show()
```

Create 3 Hierarchical Clusters

```python
hierarchical_clusters = fcluster(
    linkage_matrix,
    t=3,
    criterion='maxclust'
)

cluster_data['Hierarchical_Cluster'] = hierarchical_clusters

print(cluster_data.head())
```

Hierarchical Cluster Sizes

```python
print(
    cluster_data['Hierarchical_Cluster'].value_counts()
)
```

Compare K-Means and Hierarchical Clustering

```python
comparison = pd.crosstab(
    cluster_data['KMeans_Cluster'],
    cluster_data['Hierarchical_Cluster']
)

print(comparison)
```
